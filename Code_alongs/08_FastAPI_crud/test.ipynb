```python
import requests

response = requests.get("http://127.0.0.1:8000/books")
print(response.json())
```

```python
# KOLLA OM DET FUNKAR VIA TITEL
params = {"book_title": "Hitch"}
response = requests.get("http://127.0.0.1:8000/books/", params=params)
print(response.json())
```

```python
# KOLLA OM DET FUNKAR VIA AUTHOR OCH YEAR, FUNKAR MED PARAMETRARNA DIREKT OCKSÅ
response = requests.get("http://127.0.0.1:8000/books/", params={"book_author": "J.R.R", "book_year": 1950})
print(response.json())
```

```python
params = {
  "author": "Kokchun Giang",
  "id": 11,
  "title": "Learn with AIgineer",
  "year": 2025
}

response = requests.put("http://127.0.0.1:8000/books/1", json=params)
print(response.json())
```

```python
# DELETE OOK BY ID, SHOWING THE TITLE
response = requests.delete("http://127.0.0.1:8000/books/delete/7")
print(response.text)
```
