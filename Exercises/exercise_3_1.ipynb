### Validate data from API using Pydantic

```python
import requests

headers = {"Accept": "application/json"}
response = requests.get("https://icanhazdadjoke.com/", headers=headers)


data = response.json()
print(data)
```

### a) Create a Pydantic model with name Joke with the following fields

id with type integer
joke with type string

```python
from pydantic import BaseModel

class Joke(BaseModel):
    id: int # vi vill tydligen ha en int
    joke: str
```

### b) Validate the data from the API using the Joke model. Test out your Joke instance to see that you can access the joke and id fields.

```python
# ID är ju ingen int så det lär inte fungera
try:
    joke_instance = Joke(**data)
    print("Joke ID:", joke_instance.id)
    print("Joke text:", joke_instance.joke)
except Exception as e:
    print("Validation error:", e)
```

```python
from pydantic import computed_field

class Joke_correct(BaseModel):
    id: str
    joke: str
    
    # Computed fields - Fält som beräknas från andra fält i modellen.
    @computed_field
    @property
    def words_in_joke(self) -> int:
        """Return number of words in the joke"""
        return len(self.joke.split())
```

```python
import requests, time

headers = {"Accept": "application/json"}
response = requests.get("https://icanhazdadjoke.com/", headers=headers)

jokes_list = []

for _ in range(10):
    data = response.json()
    jokes_list.append(Joke_correct(**data))
    time.sleep(5) #för att inte få samma skämt, funkade inte ändå


print("ID: ", joke_instance.id  )
print("Joke: ", joke_instance.joke)
print("Words in joke: ", joke_instance.words_in_joke)
```
