## Dela upp X och y, X blir feature, y blir label

```python
X, y = df.drop("price", axis="columns"), df["price"]
X.head()
```

## EVALUATE

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

print(f"{mae = }")
print(f"{mse = }")
print(f"{rmse = }")
```

## Använd RidgeCV


```python
from sklearn.linear_model import RidgeCV

def evaluate_model(model, X_train, X_test, y_train, y_test):
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    
    return {"mae": mae, "mse": mse, "rmse": rmse}
```

## RidgeCV

```python
ridge = RidgeCV(alphas=[0.1, 1.0, 10.0])  # lista med olika alpha-värden som RidgeCV testar
ridge_results = evaluate_model(ridge, X_train, X_test, y_train, y_test)
print("RidgeCV:", ridge_results)
```
