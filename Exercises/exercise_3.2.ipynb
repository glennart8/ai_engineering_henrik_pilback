b) Use gemini to simulate 20 data points in json format containing the following fields: first_name, last_name, phone_number, email, department, salary, title. See if you can prompt to direct the LLM output to have swedish names, phone numbers in swedish format (+46 731 29 52), departments (IT, HR, marketing, sales), reasonable salary (you might need to check some swedish statistics on salaries) and corresponding titles within these departments.

```python
def ask_llm(promt):
    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=promt
    )
    
    return response.text

question = """
Du är en expert på företag och strukturer inom dessa.
Generera en lista av 20 data points i JSON-format innehållande dessa fält:
first_name, last_name, phone_number, email, department, salary and title. 

first_name och last_name ska vara svenska namn. phone_number ska vara i svenskt format (+46 703567867),
departments ska innehålla IT, HR, marketing och sales. salary ska vara rimlig och grundad i statistik för liknande tjänster i Sverige.
Titlarna ska vara anpassade efter respektive department.

Returnera ENDAST listan, NO backticks.
"""

response = ask_llm(question)
```

c) Now use pydantic to validate this json and put in proper schema that the fields should follow. You might need to do some processing such as removing backticks and maybe loading json data into a list with json.loads(). Also make sure that only correctly validated data should be stored.

# Parse and validate data>

```python
response
```

```python
import json

json.loads(response)
```

```python
from pydantic import BaseModel, EmailStr, Field
import json 

class Employee(BaseModel):
    first_name: str
    last_name: str
    phone_number: str
    email: EmailStr
    department: str
    salary: int = Field(gt=10000, lt = 150000) 
    title: str
    
class EmployeeList(BaseModel):
    objects: list[Employee]
    
employees = EmployeeList.model_validate({"objects": json.loads(response)})
employees
```

```python
len(employees.objects), employees.objects
```

d) Write this json data to a folder called output_data.

```python
with open("output_data/employees.json", "w") as json_file:
    json_file.write(employees.model_dump_json(indent=3))
```

e) Use pandas to read the data as dataframe

```python
import pandas as pd

df_employees = pd.DataFrame([employee.model_dump() for employee in employees.objects])
df_employees.head()
```

f) Write a csv file to your output_data


```python
df_employees.to_csv("output_data/df_employees.csv")
```

g) Load this data into a staging layer and store this into a table called employees.


```python
import dlt

@dlt.resource(write_disposition="replace", table_name="employees")
def load_data():
    yield df_employees
    
pipeline = dlt.pipeline(
    pipeline_name="employees",
    destination="duckdb", 
    dataset_name="staging"
)

load_info = pipeline.run(load_data())
print(load_info)
```

```python
departments_data = """
Du är en expert på företag och strukturer inom dessa.
Generera data i JSON-format innehållande dessa fält:
department, desciption_field och contact_person

departments ska innehålla IT, HR, marketing och sales
Alla värden i contact_person ska ha turkiska namn.

Returnera ENDAST listan, NO backticks.
"""
departments_response = ask_llm(departments_data)
departments_response
```

i) Add a departments table in your duckdb database under staging layer to store this data.

```python
import json
import dlt

departments = json.loads(departments_response)

@dlt.resource(write_disposition="replace", table_name="departments")
def load_departments():
    yield departments

departments_info = pipeline.run(load_departments())
print(departments_info)
```
